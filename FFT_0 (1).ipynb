```python
#fft excersize 0
import matplotlib.pyplot as plt
import numpy as np
from numpy.fft import fft, ifft
def Spectrum1(x, Fs):
    '''
    One-sided amplitude spectrum of x(t).
    '''

    N = len(x) # length of the signal
    frq = (float(Fs)/N)*np.arange(0, N/2+1) # one-sided frequency range
    X = fft(x)/N # fft computing and normalization
    X = X[0:N//2+1] # take just positive values
# (in python3, // ensures integer division)
# frq is array of actual frequencies
# X is complex array of amplitudes
    return frq , X
    #if an error says x is not defined its because the values are not inside the function
```

```python
def f(N, Fs): #Fs is samping rate can fix values in here also for nu t ect
    nu=250
    lam=5
    tau = N/Fs # from knwoing fs=1/deltat = N/tau can rearrange
    t = np.linspace(0, tau, N, endpoint=False) # creates N points to cover tau␣
    return np.cos(2*np.pi*nu*t)*np.exp(-lam*t)
plt.plot(f(512,1024)) #N and samping rate
```

```python
#to get it as a func of time
N=512
t = np.linspace(0, 512/1024, N, endpoint=False)
plt.plot(t,f(512,1024)) #N and samping rate
plt.xlabel('time')
plt.ylabel('N')
```

```python
frq, X= Spectrum1(f(512,1024), 1024)  #x is the data so can just take f(512 etc)
plt.plot(frq,np.abs(X), label='512')  #to get real X use abs 
frq128, X128= Spectrum1(f(128,1024), 1024)
plt.plot(frq128,np.abs(X128), label='128') 
plt.xlabel('f (Hz)')
plt.ylabel('mag X')
plt.xlabel('f (Hz)')
plt.ylabel('mag X')
plt.legend()
```
